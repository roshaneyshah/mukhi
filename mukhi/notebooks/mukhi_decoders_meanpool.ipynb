### 3. The run

Rebuilds the conditions with the normalisation fix, checks the fix took,
then runs three decoders trained and random-weight with mean pooling.
About 8.5 hours.


```python
# Press once and leave.
#
# Two fixes ride on this run, so the conditions it rebuilds are not the ones
# the earlier runs used:
#   FINDING_08  decoders were pooled at the last token, which made the
#               embedding-layer statistic a punctuation difference. Mean
#               pooling here. Output files carry a _poolmean tag.
#   FINDING_09  only generated conditions were NFC-normalised, so the
#               de-voweled control differed from Gurmukhi in Unicode form as
#               well as content. Every condition is normalised at the source
#               now, which also removes XLM-R's Gurmukhi UNK rate.
#
# Four contrasts rather than five: the Shahmukhi-Urdu language contrast is
# dropped, which also drops one condition to extract. Measured from the
# 28 September run, expect about 8.5 hours against Kaggle's 12h limit.
# Ordered so the Qwen ladder lands before Gemma if it runs long.
import subprocess, sys, time, shutil, os

CORE = ("pan_Guru|pan_Shah,pan_Guru|pan_Guru_devowel,"
        "pan_Guru_devowel|pan_Shah,srp_Cyrl|srp_Latn")
MODELS = ["Qwen/Qwen2.5-1.5B", "Qwen/Qwen2.5-3B", "google/gemma-2-2b"]

log = []


def run(label, args):
    t0 = time.time()
    print("\n" + "=" * 70 + "\n" + label, flush=True)
    p = subprocess.run([sys.executable, "run_all.py"] + args,
                       capture_output=True, text=True)
    print((p.stdout or "")[-3000:] + (p.stderr or "")[-3000:], flush=True)
    log.append((label, p.returncode, round(time.time() - t0)))
    return p.returncode == 0


def zip_now(note):
    shutil.make_archive("/kaggle/working/mukhi_results_FINAL", "zip", "results")
    print("  zip updated:", note, flush=True)


have_conditions = run("conditions", ["--stage", "conditions",
                                     "--cache-dir", CACHE, "--force"])

if have_conditions:
    import json as _json, unicodedata as _ud
    _c = _json.load(open("results/conditions_flores.json"))["conditions"]
    _bad = {k: sum(1 for s in v for ch in s
                   if ch in "\u0A33\u0A36\u0A59\u0A5A\u0A5B\u0A5E")
            for k, v in _c.items()}
    print("  precomposed nukta letters per condition:", _bad)
    assert sum(_bad.values()) == 0, "FINDING_09 fix did not take"
    print("  all conditions NFC:",
          all(_ud.normalize("NFC", s) == s for v in _c.values() for s in v))

for m in (MODELS if have_conditions else []):
    base = ["--stage", "main", "--models", m, "--contrasts", CORE,
            "--n-perm", "100", "--pooling", "mean", "--cache-dir", CACHE]
    run(m + " trained", base)
    zip_now(m + " trained")
    run(m + " random-init",
        base + ["--random-init", "--seed", "0", "--purge-cache"])
    zip_now(m + " random-init")

run("report", ["--stage", "report"])
zip_now("final")

print("\n" + "=" * 70)
for label, rc, secs in log:
    print(("ok  " if rc == 0 else "FAIL") + "  %-34s %5ds" % (label, secs))
print("\nzip:", os.path.getsize("/kaggle/working/mukhi_results_FINAL.zip") // 1024, "KB")
```
