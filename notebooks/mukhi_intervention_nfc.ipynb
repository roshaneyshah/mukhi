## 2. Hugging Face token

Needs the Kaggle secret `kaggle-mukhi` switched on **for this notebook**.
Add-ons > Secrets. A new notebook has no secrets attached, whatever your
other notebooks have.


```python
SECRET_LABEL = "kaggle-mukhi"

!pip -q install uroman 2>/dev/null
import os, torch

tok = None
try:
    from kaggle_secrets import UserSecretsClient
    tok = UserSecretsClient().get_secret(SECRET_LABEL)
except Exception as e:
    print("Could not read the Kaggle secret:", str(e)[:160])

if not tok:
    raise SystemExit(
        "No Hugging Face token.\n"
        "Secrets are attached per notebook, so a new notebook starts with "
        "none even though others have it.\n"
        "Fix: Add-ons > Secrets, switch on '" + SECRET_LABEL + "' for THIS "
        "notebook, then re-run.")

os.environ["HF_TOKEN"] = tok
os.environ["HUGGING_FACE_HUB_TOKEN"] = tok
try:
    from huggingface_hub import login
    login(token=tok, add_to_git_credential=False)
except Exception as e:
    print("login() failed, environment variables still set:", str(e)[:100])

# Preflight. dataset_info answers for gated repos WITHOUT a token, so it is
# useless as an auth check and gave a false pass on the previous run.
# Downloading a file from the gated repo does require authorisation.
from huggingface_hub import HfApi, hf_hub_download
who = HfApi().whoami(token=tok)
print("authenticated as:", who.get("name", "?"))
hf_hub_download("openlanguagedata/flores_plus", "README.md",
                repo_type="dataset", token=tok)
print("gated FLORES+ download authorised")

print("torch", torch.__version__, "| cuda", torch.cuda.is_available(),
      "| gpus", torch.cuda.device_count())
assert torch.cuda.is_available(), "Turn the GPU on: Settings > Accelerator > GPU T4 x2"
```

## 3. The run

Conditions rebuilt and checked, then the tokeniser premium and the
SIB-200 intervention. About an hour.


```python
# Press once and leave. The intervention and tokeniser repair run.
#
# FINDING_09 again. The first fix normalised the FLORES path only. SIB-200
# and PuMVR are separate entry points in run_all.py and were missed, so the
# intervention arm's de-voweled ceiling carried the same Unicode confound
# (1,194 precomposed nukta characters in SIB's Gurmukhi, none in its control).
# Both are fixed; this recomputes what depended on them.
#
# Also rebuilds the tokeniser premium, whose numbers came from a run that
# predates any of the normalisation work.
#
# GPU is needed only for the intervention. Expect about an hour.
import subprocess, sys, time, shutil, os, json, unicodedata

log = []
NUKTA = "\u0A33\u0A36\u0A59\u0A5A\u0A5B\u0A5E"


def run(label, args):
    t0 = time.time()
    print("\n" + "=" * 70 + "\n" + label, flush=True)
    p = subprocess.run([sys.executable, "run_all.py"] + args,
                       capture_output=True, text=True)
    print((p.stdout or "")[-2500:] + (p.stderr or "")[-2500:], flush=True)
    log.append((label, p.returncode, round(time.time() - t0)))
    return p.returncode == 0


def zip_now(note):
    shutil.make_archive("/kaggle/working/mukhi_results_FINAL", "zip", "results")
    print("  zip updated:", note, flush=True)


def scan(obj, path, out):
    if isinstance(obj, dict):
        for k, v in obj.items():
            scan(v, path + "/" + str(k), out)
    elif isinstance(obj, list) and obj and isinstance(obj[0], str):
        n = sum(1 for s in obj for ch in s if ch in NUKTA)
        bad = sum(1 for s in obj if unicodedata.normalize("NFC", s) != s)
        if n or bad:
            out.append((path, n, bad))


ok = run("conditions (FLORES + SIB, both normalised)",
         ["--stage", "conditions", "--cache-dir", CACHE, "--force"])

if ok:
    bad = []
    for f in ("conditions_flores.json", "conditions_sib.json"):
        scan(json.load(open("results/" + f)), f, bad)
    print("  conditions carrying precomposed nukta or non-NFC text:", bad or "none")
    assert not bad, "FINDING_09 fix did not reach every condition; stopping"

if ok:
    run("fertility (flores)", ["--stage", "fertility", "--cache-dir", CACHE])
    zip_now("fertility flores")
    run("fertility (pumvr)",
        ["--stage", "fertility", "--data", "pumvr", "--cache-dir", CACHE])
    zip_now("fertility pumvr")
    run("intervention (SIB-200)",
        ["--stage", "intervention",
         "--models", "xlm-roberta-base,google/muril-base-cased",
         "--cache-dir", CACHE])
    zip_now("intervention")

run("report", ["--stage", "report"])
zip_now("final")

print("\n" + "=" * 70)
for label, rc, secs in log:
    print(("ok  " if rc == 0 else "FAIL") + "  %-44s %5ds" % (label, secs))
print("\nzip:", os.path.getsize("/kaggle/working/mukhi_results_FINAL.zip") // 1024, "KB")
```
