## 3. The run

Three encoders. XLM-R base and MuRIL get three random-init seeds each,
XLM-R large one, matching the runs these results merge into.


```python
# Press once and leave. The encoder repair run.
#
# FINDING_09: only generated conditions were NFC-normalised, so Gurmukhi
# arrived with precomposed nukta letters that XLM-R maps to UNK while the
# de-voweled control did not. build_conditions now normalises at the source.
#
# Checked before building this: the fix changes the text of exactly one
# condition, pan_Guru. pan_Shah, pan_Guru_devowel and every romanised
# condition are byte-identical before and after. So only the two contrasts
# that involve pan_Guru need recomputing:
#     pan_Guru|pan_Shah            the headline Punjabi contrast
#     pan_Guru|pan_Guru_devowel    the information contrast
# pan_Guru_devowel|pan_Shah, Serbian, Hindi-Urdu and the romanised contrasts
# are untouched and are NOT rerun.
#
# Output carries an _nfc tag so these partial results cannot overwrite the
# full runs they will be merged into.
import subprocess, sys, time, shutil, os, json, unicodedata

PAIRS = "pan_Guru|pan_Shah,pan_Guru|pan_Guru_devowel"
PLAN = [("xlm-roberta-base", [None, 0, 1, 2]),
        ("google/muril-base-cased", [None, 0, 1, 2]),
        ("xlm-roberta-large", [None, 0])]

log = []


def run(label, args):
    t0 = time.time()
    print("\n" + "=" * 70 + "\n" + label, flush=True)
    p = subprocess.run([sys.executable, "run_all.py"] + args,
                       capture_output=True, text=True)
    print((p.stdout or "")[-2500:] + (p.stderr or "")[-2500:], flush=True)
    log.append((label, p.returncode, round(time.time() - t0)))
    return p.returncode == 0


def zip_now(note):
    shutil.make_archive("/kaggle/working/mukhi_results_FINAL", "zip", "results")
    print("  zip updated:", note, flush=True)


ok = run("conditions (rebuilt with the fix)",
         ["--stage", "conditions", "--cache-dir", CACHE, "--force"])

if ok:
    c = json.load(open("results/conditions_flores.json"))["conditions"]
    bad = sum(1 for v in c.values() for s in v
              for ch in s if ch in "\u0A33\u0A36\u0A59\u0A5A\u0A5B\u0A5E")
    nfc = all(unicodedata.normalize("NFC", s) == s
              for v in c.values() for s in v)
    print("  precomposed nukta letters remaining:", bad, "| all NFC:", nfc)
    assert bad == 0 and nfc, "FINDING_09 fix did not take; stopping"

for model, seeds in (PLAN if ok else []):
    for seed in seeds:
        base = ["--stage", "main", "--models", model, "--contrasts", PAIRS,
                "--n-perm", "200", "--out-tag", "nfc", "--cache-dir", CACHE]
        if seed is None:
            run(f"{model} trained", base)
        else:
            run(f"{model} random-init seed {seed}",
                base + ["--random-init", "--seed", str(seed)])
        zip_now(f"{model} seed={seed}")
    run("purge " + model, ["--stage", "main", "--models", model,
                           "--contrasts", PAIRS, "--out-tag", "nfc",
                           "--cache-dir", CACHE, "--purge-cache"])

zip_now("final")
print("\n" + "=" * 70)
for label, rc, secs in log:
    print(("ok  " if rc == 0 else "FAIL") + "  %-40s %5ds" % (label, secs))
print("\nzip:", os.path.getsize("/kaggle/working/mukhi_results_FINAL.zip") // 1024, "KB")
```
