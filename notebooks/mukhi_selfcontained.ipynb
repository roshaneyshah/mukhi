## 3. The run

Four models, each trained and random-weight, then the report and a zip in the
notebook Output. Expect several hours.

```python
# Press once and leave. Each model runs on its own, its weights are purged
# afterwards, and a failure is recorded rather than killing the run.
import subprocess, sys, time, shutil, os

CORE = ("pan_Guru|pan_Shah,pan_Guru|pan_Guru_devowel,"
        "pan_Guru_devowel|pan_Shah,pan_Shah|urd_Arab,srp_Cyrl|srp_Latn")
# Ordered by value per hour. The two within-family ladders come first
# (XLM-R base -> large, Qwen 1.5B -> 3B); Gemma is a singleton and goes last,
# so a timeout costs the least important model. Measured cost: about 9h total
# against Kaggle's 12h commit limit.
BIG = ["xlm-roberta-large", "Qwen/Qwen2.5-1.5B",
       "Qwen/Qwen2.5-3B", "google/gemma-2-2b"]

log = []


def run(label, args):
    t0 = time.time()
    print("\n" + "=" * 70 + "\n" + label, flush=True)
    p = subprocess.run([sys.executable, "run_all.py"] + args,
                       capture_output=True, text=True)
    out = (p.stdout or "")[-3000:] + (p.stderr or "")[-3000:]
    print(out, flush=True)
    log.append((label, p.returncode, round(time.time() - t0)))
    return p.returncode == 0


# Conditions come from Hugging Face; without them nothing downstream can run,
# so a failure here skips the models rather than burning GPU hours on nothing.
have_conditions = run("conditions", ["--stage", "conditions",
                                     "--cache-dir", CACHE])

for m in (BIG if have_conditions else []):
    base = ["--stage", "main", "--models", m, "--contrasts", CORE,
            "--n-perm", "100", "--cache-dir", CACHE]
    run(m + " trained", base)
    # purge only after the random-init pass, which reuses the same config
    run(m + " random-init",
        base + ["--random-init", "--seed", "0", "--purge-cache"])
    # Re-zip after every model, so a timeout still leaves usable output.
    shutil.make_archive("/kaggle/working/mukhi_results_FINAL", "zip", "results")
    print("  zip updated after", m, flush=True)

run("report", ["--stage", "report"])

print("\n" + "=" * 70)
for label, rc, secs in log:
    print(("ok  " if rc == 0 else "FAIL") + "  %-34s %5ds" % (label, secs))

shutil.make_archive("/kaggle/working/mukhi_results_FINAL", "zip", "results")
print("\nzip written:",
      os.path.getsize("/kaggle/working/mukhi_results_FINAL.zip") // 1024, "KB")
```
